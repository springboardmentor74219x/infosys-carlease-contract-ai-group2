```python
import pandas as pd
import numpy as np
df=pd.read_csv('../notebook/Milestone1_pavan-srinivas.csv')
df.head()
```

```python
#Ground Truth Simulation

df['expected_apr']=df['apr']
df['expected term']=df['term_months']
df['expected_payment']=df['monthly_payment']
df['expected_penalty']=df['penalty_clause']
```

```python
df.columns 

df["expected_penalty"] = df["expected_penalty"] 
df[["expected_penalty"]].head()
```

```python
# if extracted APR exactly matches expected
# APR score 1
# if not score 0

# if extracted term months equals expected
# term months score 1
# else score 0

# if extracted penalty text matches expected
# penalty score 1
# else score 0

# Create expected columns from existing ground-truth values
df["expected_apr"] = df["apr"]
df["expected_term"] = df["term_months"]
df["expected_penalty"] = df["penalty_clause"]

# Scoring logic
df["apr_score"] = (df["apr"] == df["expected_apr"]).astype(int)
df["term_score"] = (df["term_months"] == df["expected_term"]).astype(int)
df["penalty_score"] = (df["penalty_clause"] == df["expected_penalty"]).astype(int)

# Total score
df["total_score"] = (
    df["apr_score"] +
    df["term_score"] +
    df["penalty_score"]
)
```
