```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/sample_car_contracts.csv')
df.head()
```

```python
# BAR CHART 1 
# Bar chart: count of customers in each risk category (low, medium, high)

risk_counts = df["risk_flag"].value_counts()

plt.figure(figsize=(6,4))
risk_counts.plot(kind="bar")
plt.title("Number of Customers by Risk Flag")
plt.xlabel("Risk Category")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)   
plt.tight_layout()
plt.show()
```

```python
#  Bar chart 2 
# Bar chart: how many customers have Car Loan vs Car Lease

contract_counts = df["contract_type"].value_counts()

plt.figure(figsize=(6,4))
contract_counts.plot(kind="bar")
plt.title("Number of Customers by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
