```python
#TASK1 
#Loading and Reading the Dataset
#Finding and Cleaning the Missing Values
#Filtering Based on Category
import pandas as pd
```

```python

df=pd.read_csv('../data/sample_car_contracts.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
numeric=df.select_dtypes(include='number').columns
categorical=df.select_dtypes(include='object').columns
df[numeric]=df[numeric].fillna(df[numeric].mean())
df[categorical]=df[categorical].fillna(df[categorical].mode().iloc[0])
```

```python
suv = df[df['vehicle_type'] == "SUV"]
suv
```

```python
suv_sedan = df[df['vehicle_type'].isin(["SUV", "Sedan"])]
suv_sedan
```

```python
emi_above_20k = df[df['monthly_emi'] > 20000]
emi_above_20k
```

```python
suv_emi_above_25k = df[(df['vehicle_type'] == "SUV") & (df['monthly_emi'] > 25000)]
suv_emi_above_25k
```
