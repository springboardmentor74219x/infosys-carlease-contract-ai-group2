```python
import pandas as pd

df = pd.read_csv(r'C:\Users\sriha\Downloads\sample (1).csv')
```

```python
import matplotlib.pyplot as plt
risk_counts = df['risk_flag'].value_counts()
plt.figure(figsize=(8,5))
plt.bar(risk_counts.index.astype(str), risk_counts.values)
plt.xlabel("Risk Category")
plt.ylabel("Number of Customers")
plt.title("Customer Count by Risk Flag")
plt.show()
```

```python
avg_interest = df.groupby('risk_flag')['interest_rate'].mean()
plt.figure(figsize=(8,5))
plt.bar(avg_interest.index.astype(str), avg_interest.values)
plt.xlabel("Risk Category")
plt.ylabel("Average Interest Rate (%)")
plt.title("Average Interest Rate by Risk Category")
plt.show()
```
