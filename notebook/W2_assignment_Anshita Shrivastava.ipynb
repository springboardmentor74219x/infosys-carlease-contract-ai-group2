```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("../data/sample_car_contracts.csv")
df.head()
```

```python
#1)Count of Customers per car type
plt.figure(figsize=(8,5))
sns.countplot(data=df,x='vehicle_type')
plt.title("Number of Customers per Car Type")
plt.xlabel("Car Type")
plt.ylabel("Count")
#plt.show()
#insight 
plt.figtext(0.5, -0.1,"Insight: SUVs and Sedan are the most common car type, while EV  have the fewest customers.", ha="center", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
#2)Average EMI per Car Type 
plt.figure(figsize=(8,5))
df.groupby('vehicle_type')['monthly_emi'].mean().plot(kind='bar',color='green')
plt.title("Average Monthly EMI per Car Type")
plt.xlabel("Car Type")
plt.ylabel("Aerage EMI")
#plt.show()
#insight
plt.figtext(0.5, -0.1,"Insight: SUVs have the highest avergae EMI , while Hatchbacks  have the lowest.", ha="center", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
#3)Risk Flag Distribution
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="risk_flag", hue="risk_flag", palette="viridis", legend=False)
plt.title("Risk Flag Distribution")
plt.xlabel("Risk Category")
plt.ylabel("Number Of Customers")
#plt.show()
#insight
plt.figtext(0.5, -0.1,"Insight: High risk customers are more frequent compared to low frequent ones.", ha="center", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
#4) Distribution of Contract Types
plt.figure(figsize=(7,7))
df["contract_type"].value_counts().plot.pie(autopct="%1.1f%%",startangle=90, colors=["#4c72b0", "#55a868"])
plt.title("Distribution of Contract Types (Loan vs Lease)")
plt.ylabel("")
#plt.show()
#insight
plt.figtext(0.5, -0.1,"Insight:Loan and Lease customers are equal.", ha="center", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
#5)Tenure Distribution
plt.figure(figsize=(8,5))
plt.hist(df['tenure_months'],bins=7,edgecolor='black')
plt.title("Distribution of Tenure(Months)")
plt.xlabel("Tenure(Months)")
plt.ylabel("Frequency")
#plt.show()
#insight
plt.figtext(0.5, -0.1,"Insight: Most Customers have tenure between 30-50 months , only a few extended it beyond 60 months", ha="center", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
#BONUS TASK 
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="tenure_months",y="monthly_emi",hue="risk_flag",palette="viridis",s=120)
plt.title("Scatter Plot of Tenure vs Monthly EMI")
plt.xlabel("Tenure(Months)")
plt.ylabel("Monthly EMI")
plt.legend(title="Risk Level")
plt.grid(True,linestyle='--',alpha=0.3)
#plt.show()
#insight
plt.figtext(0.5, -0.1,"Insight: Customers with high risk tend to have higher emi and longer tenure", ha="center", fontsize=12)
plt.tight_layout()
plt.show()
```
