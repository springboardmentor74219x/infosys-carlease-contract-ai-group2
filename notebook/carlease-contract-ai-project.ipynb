# Data Analysis of Car Contracts
This notebook analyzes the car contracts dataset to extract insights and perform evaluations.

```python
# Load the dataset
# Uncomment the appropriate line based on the file location
# df = pd.read_csv('../data/sample_car_contract.csv')
# df = pd.read_csv('C:/Users/Admin/Downloads/sample_car_contracts.csv')
df = pd.read_csv('../data/sample_car_contracts.csv')
df.head()  # Display the first few rows of the dataset
```

```python
# Display information about the dataset
df.info()
```

```python
# Get descriptive statistics of the dataset
df.describe()
```

```python
# Analyze the distribution of contract types and risk flags
print("Contract type counts:")
print(df['contract_type'].value_counts())

print("\nRisk flag counts:")
print(df['risk_flag'].value_counts())

# Filter and display high-risk records
high_risk = df[df['risk_flag'].str.lower() == 'high']
print("\nHigh-risk records:")
high_risk
```

```python
# Analyze value counts for various columns
print("Contract Type Value counts \n",df["contract_type"].value_counts())

print("\n Veichle Type Value counts \n",df["vehicle_type"].value_counts())

print("\n Interet Type Value counts \n",df["interest_rate"].value_counts())
```

```python
# Analyze the average interest rate based on risk flag
print("Average of the risk flag " , df.groupby('risk_flag')['interest_rate'].mean())
```
