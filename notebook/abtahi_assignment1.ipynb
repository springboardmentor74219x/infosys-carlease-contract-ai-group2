```python
import matplotlib.pyplot as plt
risk_counts = df['risk_flag'].value_counts()
plt.figure(figsize=(8,5))
plt.bar(risk_counts.index.astype(str), risk_counts.values)
plt.xlabel("Risk Category")
plt.ylabel("Number of Customers")
plt.title("Customer Count by Risk Flag")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Asus\Desktop\infosys\infosys-carlease-contract-ai-group2\data\sample_car_contracts.csv")

# Use the correct text column name
text_col = "contract"   # CHANGE THIS IF your column is different

# 1️⃣ Contract length
df["contract_length"] = df[text_col].apply(lambda x: len(str(x).split()))

# 2️⃣ Hidden charges
keywords = ["fee", "penalty", "late", "extra"]

def check_hidden(text):
    text = str(text).lower()
    return any(word in text for word in keywords)

df["hidden_charges_flag"] = df[text_col].apply(check_hidden)

# 3️⃣ Extract high-risk words
high_risk_text = " ".join(
    df[df["risk_category"] == "high"][text_col].astype(str).tolist()
).lower()

from collections import Counter

stopwords = {"the","and","is","to","of","in","a","for","on","with","by","an"}
words = [word for word in high_risk_text.split() if word not in stopwords]
word_freq = Counter(words)
top_keywords = word_freq.most_common(10)
top_keywords
```
