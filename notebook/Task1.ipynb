```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("../data/sample_car_contracts.csv")
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='risk_flag')
plt.xlabel("Risk Category")
plt.ylabel("Number of Customers")
plt.title("Customer Count by Risk Level")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1) Monthly EMI per Customer
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="customer_name", y="monthly_emi", palette="Blues")
plt.title("Monthly EMI for Each Customer")
plt.xlabel("Customer")
plt.ylabel("Monthly EMI")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 2) Interest Rate per Customer
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="customer_name", y="interest_rate", palette="Greens")
plt.title("Interest Rate for Each Customer")
plt.xlabel("Customer")
plt.ylabel("Interest Rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 3) Tenure Months per Customer
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="customer_name", y="tenure_months", palette="Oranges")
plt.title("Tenure (Months) for Each Customer")
plt.xlabel("Customer")
plt.ylabel("Tenure (Months)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 4) Issue Type Distribution
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="issue_type", palette="Purples")
plt.title("Count of Each Issue Type")
plt.xlabel("Issue Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
# 5) Risk Level vs Number of Issues
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="risk_flag", hue="issue_type", palette="viridis")
plt.title("Risk Level vs Issue Frequency")
plt.xlabel("Risk Category")
plt.ylabel("Issue Count")
plt.legend(title="Issue Type", bbox_to_anchor=(1.05,1), loc="upper left")
plt.tight_layout()
plt.show()
```
