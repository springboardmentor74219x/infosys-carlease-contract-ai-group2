```python
df = pd.read_csv("../data/contract_evaluation_dataset_tulasi-vasireddy.csv")
df.head()
```

```python
sample_df = df.head(5)
sample_df
```

```python
def extract_sla(contract_text):
    return {
        "apr": None,
        "term_months": None,
        "monthly_payment": None,
        "penalty_clause": None
    }
```

```python
results = [extract_sla(text) for text in sample_df["contract_text"]]
results_df = pd.DataFrame(results)
results_df
```

```python
comparison_df = sample_df.reset_index(drop=True)

comparison_df["apr_match"] = (
    comparison_df["expected_apr"] == results_df["apr"]
).astype(int)

comparison_df["term_match"] = (
    comparison_df["expected_term"] == results_df["term_months"]
).astype(int)

comparison_df["payment_match"] = (
    comparison_df["expected_payment"] == results_df["monthly_payment"]
).astype(int)

comparison_df["penalty_match"] = (
    comparison_df["expected_penalty"] == results_df["penalty_clause"]
).astype(int)

comparison_df
```
