```python
import pandas as pd
df = pd.read_csv('../data/sample_car_contracts.csv')
```

*Missing values*

```python
print("Missing values per column:")
print(df.isnull().sum())
```

*Filtering*

```python
df['risk_flag'] = df['risk_flag'].str.lower()

df = df[df['monthly_emi'] > 0]

df = df[(df['tenure_months'] >= 12)]

df = df[df['interest_rate'] >= 0]

df.head(13)
```

*Filter by risk level*

```python
high_risk = df[df['risk_flag'] == 'high']
print(f"\nHigh risk contracts: {len(high_risk)}")
high_risk[['customer_name', 'contract_type', 'risk_flag', 'issue_type']].head(13)
```
