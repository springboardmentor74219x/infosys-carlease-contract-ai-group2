```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('../data/sample_car_contracts.csv')
```

```python
plt.figure(figsize=(6,4))
df['risk_flag'].value_counts().plot(kind='bar')
plt.xlabel("Risk Flag")
plt.ylabel("Number of Customers")
plt.title("Customers by Risk Level")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
df['vehicle_type'].value_counts().plot(kind='bar')
plt.xlabel("Vehicle Type")
plt.ylabel("Number of Customers")
plt.title("Customers by Vehicle Type")
plt.tight_layout()
plt.show()
```

```python
plt.scatter(df['tenure_months'], df['monthly_emi'])
plt.xlabel("Tenure (Months)")
plt.ylabel("Monthly EMI")
plt.title("Tenure vs Monthly EMI")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
avg_interest = df.groupby('risk_flag')['interest_rate'].mean()
```

```python
print("----- Insights -----")
print("-"*60)

print(f"1. High-risk customers pay an average interest rate of {avg_interest.get('high', 0):.2f}%.")
print(f"2. Medium-risk customers pay an average interest rate of {avg_interest.get('medium', 0):.2f}%.")
print(f"3. Low-risk customers pay an average interest rate of {avg_interest.get('low', 0):.2f}%.\n")
```
