# Contract Data Extraction & Cleaning

### STEP 1: Import Required Libraries

```python
import pandas as pd
import re
from pathlib import Path
```

### STEP 2: Load Dataset

```python
df=pd.read_csv('../data/sample_emails.csv')

df.head()
```

## STEP 3: View Columns

```python
df.columns
```

### STEP 4: Cleaning Phone Numbers

Phone numbers often contain spaces, dashes, or extra characters.
We remove everything except digits.

Example: "987-654 3210" → "9876543210"

```python
def clean_phone(p):
    return re.sub(r"\D", "", str(p))

# Create a sample phone column for demonstration since it doesn't exist in emails data
df["phone_no_sample"] = "555-123-4567"
df["phone_no_clean"] = df["phone_no_sample"].apply(clean_phone)
df["phone_no_clean"].head()
```

### STEP 5: Cleaning Email IDs

```python
df["email_id_clean"] = df["sender"].str.lower()
df["email_id_clean"].head()
```

### STEP 6: Cleaning Text Columns

```python
def clean_text(s):
    if pd.isna(s):
        return ""
    s = str(s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

df["clause_summary_clean"] = df["body"].apply(clean_text)
df["clause_summary_clean"].head()
```

We convert EMI, interest rate, and tenure into numeric types to avoid errors in analysis.

```python
# Create sample numeric columns for demonstration (not in original email data)
df["monthly_emi"] = pd.Series([5000.0, 6000.0, 5500.0, 7000.0, 6500.0] + [0.0] * (len(df) - 5))
df["interest_rate"] = pd.Series([8.5, 9.2, 8.8, 9.5, 8.0] + [0.0] * (len(df) - 5))
df["tenure_months"] = pd.Series([60, 48, 72, 60, 48] + [0] * (len(df) - 5))

df[["monthly_emi", "interest_rate", "tenure_months"]].head()
```

### STEP 8: Show Cleaned Columns

```python
cleaned_df = df[[
    "id",
    "sender",
    "email_id_clean",
    "phone_no_clean",
    "priority",
    "label",
    "monthly_emi",
    "interest_rate",
    "tenure_months",
    "clause_summary_clean"
]]

cleaned_df.head()
```

### STEP 9: Save Cleaned Dataset


```python
output_path = Path("../data/sample_emails_cleaned.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
cleaned_df.to_csv(output_path, index=False)

print("Cleaned file saved to:", output_path.resolve())
```
