```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('milestone1_jyothsnasai.csv')
df.head()
```

```python
#Ground Truth simulation
df["expected_apr"] = df["apr"]
df["expected_term_months"] = df["term_months"]
df["expected_monthly_payment"] = df["monthly_payment"]
df["expected_penalty"] = df["penalty_clause"]
```

```python
df = df.rename(columns={"penalty":"penalty_clause"})
```

```python
df.columns
```

```python
df[["penalty_clause", "expected_penalty"]].head()
```

```python
# if extracted APR exactly matches expected
# APR Score 1
#if not score 0

# if extracted term months equals  expected
# term months Score 1
#else score 0

# if extracted penalty text matches expected
# penalty Score 1
# else score 0

df["apr_score"] = (df["apr"] ==df["expected_apr"]).astype(int)
df["term_months_score"] = (df["term_months"] ==df["expected_term_months"]).astype(int)
df["payment_score"] = (df["monthly_payment"] ==df["expected_monthly_payment"]).astype(int)
df["penalty_clause_score"] = (df["penalty_clause"] ==df["expected_penalty"]).astype(int)

df["total_score"] = df["apr_score"] + df["term_months_score"] + df["payment_score"] + df["penalty_clause_score"]
df["total_percentage"] = (df["total_score"] / 4) * 100
df[["apr_score", "term_months_score", "payment_score", "penalty_clause_score", "total_score", "total_percentage"]].head()
```

```python
df.to_csv("../data/milestone2_jyothsna.csv",index=False)
```
