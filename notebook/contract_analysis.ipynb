```python
import pandas as pd
df = pd.read_csv('../data/sample_car_contracts.csv')
df.head()
```

```python
df['total_amount'] = df['monthly_emi'] * df['tenure_months']
df[['customer_name', 'monthly_emi', 'tenure_months', 'total_amount']]
```

```python
df['tenure_years'] = df['tenure_months'] / 12
avg_tenure = df.groupby('risk_flag')['tenure_years'].mean()
print(avg_tenure)
```

```python
df["total_amount"] = df["monthly_emi"] * df["tenure_months"]
print("Total Amount Added:")
print(df[["customer_name", "total_amount"]])
print("\n")

df["tenure_years"] = df["tenure_months"] / 12
print("Tenure in Years:")
print(df[["customer_name", "tenure_years"]])
print("\n")

avg_total = df.groupby("risk_flag")["total_amount"].mean()

print("Average Total Contract Amount per Risk Flag:")
print(avg_total)
print("\n")

avg_tenure = df.groupby("risk_flag")["tenure_years"].mean()
print("Average Tenure (Years) per Risk Flag:")
print(avg_tenure)

pd.crosstab(df['risk_flag'], df['recommended_action'])
```
