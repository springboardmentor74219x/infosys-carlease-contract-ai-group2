```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\DELL\Downloads\infosys-carlease-contract-ai-group2-main\infosys-carlease-contract-ai-group2\data\sample.csv")

df.head()
```

```python
df['contract_type'].value_counts().plot(kind='bar')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.title('Customers by Contract Type')
plt.show()
```

```python
df['risk_flag'].value_counts().plot(kind='bar')
plt.xlabel('Risk Flag')
plt.ylabel('Number of Customers')
plt.title('Customers by Risk Category')
plt.show()
```

```python
avg_interest = df.groupby('risk_flag')['interest_rate'].mean()
avg_interest
```

```python
colors = {'low':'green','medium':'orange','high':'red'}
plt.scatter(df['tenure_months'], df['monthly_emi'], 
            c=df['risk_flag'].map(colors))
plt.xlabel('Tenure (months)')
plt.ylabel('Monthly EMI')
plt.title('Tenure vs EMI Colored by Risk')
plt.show()

```
