```python
import pandas as pd
import matplotlib.pyplot as plt
import re
from collections import Counter
import string
file_path = r"C:\Users\DELL\Downloads\infosys-carlease-contract-ai-group2-main\infosys-carlease-contract-ai-group2\data\sample.csv"
try:
    pd.read_csv(file_path, nrows=1)
except Exception:
    file_path = '/mnt/data/9cf416e6-db00-472a-bab2-0e5e2b1cf892.csv'
df = pd.read_csv(file_path)
df.head()
```

```python
keywords = ['fee','penalty','late','extra']
pattern = re.compile(r'\b(' + '|'.join(re.escape(k) for k in keywords) + r')\b', flags=re.IGNORECASE)
df['clause_summary'] = df['clause_summary'].fillna('')
df['hidden_charges_flag'] = df['clause_summary'].apply(lambda t: 1 if pattern.search(t) else 0)
df[['id','clause_summary','hidden_charges_flag']].head(10)
```

```python
df['hidden_charges_flag'].value_counts()
```

```python
high_text = " ".join(df[df['risk_flag']=='high']['clause_summary'].fillna('').astype(str).tolist()).lower()
trans = str.maketrans(string.punctuation, ' '*len(string.punctuation))
high_text_clean = high_text.translate(trans)
tokens = [w for w in high_text_clean.split() if len(w)>2]
stopwords = {'the','and','for','all','with','that','this','from','are','must','pay','per','month','months','rate','interest','amount','be','to','of','in','on','is','a','an'}
tokens = [w for w in tokens if w not in stopwords]
counts = Counter(tokens)
top10 = counts.most_common(10)
top10
```

```python
words, freqs = zip(*top10) if top10 else ([],[])
plt.figure(figsize=(8,5))
plt.bar(words, freqs)
plt.xlabel('Keyword')
plt.ylabel('Frequency')
plt.title('Top Keywords in High-Risk Contract Clauses')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
df['risk_flag'].value_counts().plot(kind='bar')
plt.xlabel('Risk Flag')
plt.ylabel('Number of Contracts')
plt.title('Risk Distribution')
plt.tight_layout()
plt.show()
```
