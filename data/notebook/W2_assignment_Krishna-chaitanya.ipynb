```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv('../notebook/sample_car_contracts.csv')
```

```python
#Count of Customers by Contract Type
contract_counts = df['contract_type'].value_counts()
plt.figure(figsize=(6,4))
contract_counts.plot(kind='bar')
plt.title("Number of Customers by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Count of Customers")
plt.show()
```

```python
#Average Interest Rate by Risk Level
avg_interest = df.groupby('risk_flag')['interest_rate'].mean()
plt.figure(figsize=(6,4))
avg_interest.plot(kind='bar')
plt.title("Average Interest Rate per Risk Level")
plt.xlabel("Risk Level")
plt.ylabel("Average Interest Rate (%)")
plt.show()
```

```python
#Tenure vs Monthly EMI colored by Risk Level
plt.figure(figsize=(7,5))
sns.scatterplot(
    data=df,
    x='tenure_months',
    y='monthly_emi',
    hue='risk_flag'
)
plt.title("Tenure vs Monthly EMI (Colored by Risk Level)")
plt.xlabel("Tenure (months)")
plt.ylabel("Monthly EMI")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
df = pd.read_csv("../notebook/sample_car_contracts.csv")
#Add contract_length (word count)
df['contract_length'] = df['clause_summary'].apply(lambda x: len(str(x).split()))
# Detect hidden charges
# Check for: fee, penalty, late, extra, charge, interest

keywords = ["fee", "penalty", "late", "extra", "charge", "interest"]

df['hidden_charges_flag'] = df['clause_summary'].str.lower().apply(
    lambda text: any(word in text for word in keywords)
)
# Extract frequent words only in High-Risk contracts

high_risk_text = " ".join(df[df['risk_flag'] == "high"]['clause_summary'].str.lower())
word_list = high_risk_text.split()

word_freq = Counter([word for word in word_list if word in keywords])

# Visualization: Bar chart of frequent risky keywords
plt.figure()
plt.bar(word_freq.keys(), word_freq.values())
plt.title("Common Risk Keywords in High-Risk Contracts")
plt.xlabel("Keywords")
plt.ylabel("Frequency")
plt.show()
# Show Final Results

print("\nProcessed Summary Table:\n")
print(df[['customer_name','risk_flag','contract_length','hidden_charges_flag']])
print("\nFrequent Keywords in High-Risk Contracts:\n")
print(word_freq)
```
