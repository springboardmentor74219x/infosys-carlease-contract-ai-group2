```python
#Missing Values
print("Missing values column:")
print(df.isnull().sum())
print(f"\nTotal missing values: {df.isnull().sum().sum()}")
print(f"\nRows with missing value: {df.isnull().any(axis=1).sum()}")
```

```python
#Clean Missing Values
df = df.fillna({
    'monthly_emi': df['monthly_emi'].median(),
    'interest_rate': df['interest_rate'].median(),
    'tenure_months': df['tenure_months'].median(),
    'customer_name': 'Unknown',
    'risk_flag': 'medium'
})
print("After cleaning:")
print(df.isnull().sum())
```

```python
#Filter Values 
# Normalize risk_flag
df['risk_flag'] = df['risk_flag'].str.lower()
# Filter valid EMI and tenure
df = df[df['monthly_emi'] > 0]
df = df[(df['tenure_months'] >= 12) & (df['tenure_months'] <= 84)]
# Filter valid interest rates
df = df[df['interest_rate'] >= 0]
print(f"Rows after filtering: {len(df)}")
print(df.head())
```

```python
# Filter by risk level
high_risk = df[df['risk_flag'] == 'high']
print(f"\nhigh risk contracts: {len(high_risk)}")
print(high_risk[['customer_name', 'contract_type', 'risk_flag', 'issue_type']])
```
